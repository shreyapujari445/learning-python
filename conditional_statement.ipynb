```python
# score in three subjects: maths,scienceand language (out of 100)
# output:
#"eligible for honors"
#"need remedial"
#"standard placement"

maths=int(input("enter score: "))
science=int(input("enter score: "))
language=int(input("enter score: "))
avg= (maths+science+language)/3
if avg >= 85 and maths >= 70 and science >= 70 and language >= 70:
    print("eligible for honors")
elif maths < 50 or science < 50 or language < 50:
    print("remedial")
else:
    print("standard placement")

```

```python
# wap to check weather a given year is leap year or not?
year=int(input("enter any number: "))
if year%4==0:
    if year%100==0:
        if year%400==0:
            print(year,"leap year")           #century
        else:
            print(year,"not a leap year")     #century
    else:
        print(year,"leap year")               #non-century
else:
    print(year,"not a leap year")             #non-century
```
