```python
#wap to check weather a given number is an armstrong or not??
n=int(input("enter any number: "))
a=len(str(n))
sumdigit=0
old=n
while n!=0:
    dig=n%10
    sumdigit=sumdigit+(dig**a)
    n=n//10
if sumdigit==old:
    print(old,"armstrong number")
else:
    print(old,"not a armstrong number")
```
