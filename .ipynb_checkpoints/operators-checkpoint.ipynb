# operators


### arithematic operator + ,- , * , / , // , % , **

```python
a=10
b=2
print("addition:-",a+b)
print("subtraction:-",a-b)
print("multiply:-",a*b)
print("simple division:-",a/b)
print("floor division:-",a//b)
print("modulus:-",a%b)
print("exponential power:-",a**b)
```

```python
# comparision operator == ,!=,>=,<=,>,<
a=10
b=8
print(a==b)
print(a!=b)
print(a>=b)
print(a<=b)
print(a<b)
print(a>b)
```

```python
# logical operator -- and ,or , not 
a=10
b=8
print(a>b and b>c)
print(a>b or b>c)
print(a>b and not b>c)
```

```python
# identity operator - is ,is not
# id ()-- it is a predefined function that return memory adrress of given object.
a=10
b=8
print(id(a))
print(id(b))
print (a is b)
print (a is not b)
```

```python
c=20
d=20

print(id(c))
print(id(d))
print (c is d)
print (c is not d)
```

```python
#membership orperator-- in ,not in
a="hello"
print("h" in a)
print("h" not in a)
print("O" in a)
```

```python
# bitwise operator--&,|,^
a=10
b=8
print(a&b)
print(a|b)
print(a^b)
```

```python
# bin()-- it is a predefined function always return binary of give integer object.
a=10
b=8
print(bin(a))
print(bin(b))
print("------")
print(bin(a&b))
print(a&b)
print("---*---")
print(bin(a))
print(bin(b))
print("------")
print(bin(a|b))
print(a|b)
print("----")
print(bin(a))      
print(bin(b))
```
