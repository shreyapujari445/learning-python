```python
# index-position of a single character in a given string
# first character of any string is it 0 index
# last charcter is at -1 index.
# indexing- it is a technique for accessing individual character from given string.
s="welcome to bhopal"
print(s[0])
print(s[-1])
print(s[7])
print(s[-4])
```

```python
s="welcome to bhopal"
print(s[::])
print(s[:7:])
print(s[::-1])
print(s[-1:-7:-1])
```

```python
#iteration -- 
# direct iteration 
s="hello"
for i in s:
    print(i)
```

```python
# indirect iteration- with help of range function
# len()- no of character in a given string.
s="hello"
for i in range (len(s)):
    print(i,s[i])
```

```python
#wap to print no.of vowels ,no.of conconent and the no.of spaces in a given string. 
s="welcome to city of lakes"
c_v=0
c_c=0
c_s=0
for i in s:
    if i in 'aeiouAEIOU':
        c_v=c_v+1
    elif i==' ':
        c_s=c_s+1
    else :
        c_c=c_c+1
print("count of vowel",c_v)
print("count of consonent",c_c)
print("count of spaces",c_s)
```

```python
# wap to print reverse of any given string .
s=input("enter any string: ")
rev=''
for i in s:
    rev=i+rev
print(rev)             #by logic
print(s[::-1])         # by slicing
```

```python
#wap to check weather a given string is palindrome string or not.
s=input("enter any string: ")
rev=''
for i in s:
    rev=i+rev
if s==rev:
    print(s,"palindrome string")
else:
    print(s,"not a palindrome string")
```

```python
#count()-return occurence of any character in a given string.
#strobj.count(char/substring,[start],[stop])
#start- by default 0
#stop- by default end of the string
#step--- n-1
s="welcome to city of lakes"
print(s.count("o"))
print(s.count("o",5))
print(s.count(" "))
print(s.count("bhopal"))
print(s.count("z"))
print(s.count("o",5,9))
```

```python
#index -- return position of character /substring from a given string.
#str.index(char/substring,[start],[stop])
#always return first occurence.
s="welcome to bhopal"
print(s.index("o"))
print(s.index("o",5))
print(s.index("bhopal"))
print(s.index("o",5,9))    (it shows error)
```

```python
#find -- return position of character /substring from a given string.
#str.find(char/substring,[start],[stop])
#always return first occurence.
s="welcome to bhopal"
print(s.find("o"))
print(s.find("o",5))
print(s.find("bhopal"))
print(s.find("o",5,9))     --(it gives -1 on not found the substring)
```
