```python
#wap to print reverse of any given no.
n=int(input("enter any number: "))
rev=0
while n!=0:
    dig=n%10            #lastdigit
    rev=rev*10+dig      #output
    n=n//10             #loop
print(rev)
```

```python
#wap to print reverse no. between start and end take start and end as a user input at run time.


start = int(input("Enter start number: "))
end = int(input("Enter end number: "))
for n in range(start,end+1):
    rev=0
    while n!=0:
        dig=n%10
        rev=rev*10+dig
        n=n//10
    print(rev)
```
