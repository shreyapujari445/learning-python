```python
#1. Positive, Negative, or Zero
#Explanation:
#Accept a number and determine whether it is positive, negative, or zero.
#Concepts:
#- if, elif, else
#- Comparison operators.

num=int(input("enter any number: "))
if num > 0 :
    print(num,"positive")
elif num < 0 :
    print(num,"negative")
else :
    print(num,"zero")
```

```python
#2. Even or Odd
#Explanation:
#Check whether a number is divisible by 2.
#Concepts:
#- Modulus (%)
#- if...else
num=int(input("enter any number: "))
if num%2==0:
    print("even number")
else:
    print("odd number")
```

```python

#Multiple of 3 and 7
#Explanation:
#Check whether a number is divisible by both 3 and 7.
#Concepts:
#- and
#- Modulus
num=int(input("enter any number: "))

if num % 3 == 0 and num % 7 == 0:
    print(num,"divisible by 3 and 7")
else:
    print(num,"not divisible by 3 and 7")
```

```python
#Vowel or Consonant
#Explanation:
#Determine whether the entered alphabet is a vowel or consonant.
#Concepts:
#- in operator

alphabet=input("enter any alphabet: ")

if alphabet.lower() in "aeiou" :
    print(alphabet ,"is vowel")
else:
    print(alphabet ,"consonent")
```

```python
#Check Alphabet
#Explanation:
#Accept a character and determine whether it is an alphabet.
#Concepts:
#- Character comparison

character= input("enter a character: ")
if character >="a" and character <="z" or character >="A" and character <="Z":
    print(character,"is alphabet")
else:
    print(character,"not a alphabet")
```

```python

# Pass or Fail
#Explanation:
#Display Pass if marks are 35 or above, otherwise Fail.
#Concepts:
#- Comparision

marks=int(input("enter marks: "))
if marks>=35:
    print("pass")
else:
    print("fail")
```
