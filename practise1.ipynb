```python
#1. Find the Largest of Three Numbers Write a Python program that accepts three numbers 
#from the user and displays the largest number among them. 
#INPUT REQUIRED: Input required: Three numbers. 
#EXPECTED OUTPUT: Expected output: The largest number.

number1=int(input("enter any number1: "))
number2=int(input("enter any number2: "))
number3=int(input("enter any number3: "))
if number1 >number2 :
    print(number1,"the largest number")
elif number2 >number3 :
    print(number2,"the largest number")
else:
    print(number3,"the largest number")
```

```python
#2. Check Whether a Year Is a Leap Year Write a Python program that accepts a year 
#from the user and checks whether it is a leap year. A year is a leap year 
#if it is divisible by 400, or if it is divisible by 4 but not by 100.
#INPUT REQUIRED: Input required: A year (integer).
#EXPECTED OUTPUT: Expected output: Display whether the year is a leap year or not.

year=int(input("enter year: "))
if year%4==0:
    if year%100==0:
        if year%400==0:
            print(year,"leap year")           
        else:
            print(year,"not a leap year")     
    else:
        print(year,"leap year")              
else:
    print(year,"not a leap year")
        
```

```python
#3. Check Whether a Number Is a Multiple of Both 3 and 7 Write a Python program that accepts an integer and 
#checks whether it is exactly divisible by both 3 and 7. 
#INPUT REQUIRED: Input required: One integer. EXPECTED OUTPUT: 
#Expected output: Display whether the number is divisible by both 3 and 7.

integer=int(input("enter any number: "))
if integer % 3 == 0 and integer % 7 == 0:
    print(integer,"divisible by 3 and 7")
else:
    print(integer,"not divisible by 3 and 7")
```
