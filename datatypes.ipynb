```python
# datatype-- datatype in python defines the type of value stored in a variable.

#on the  basis of mutability --
# they are classify into:
#1. mutable--list, dictionary,set.
#2. immutable-- int , float,string,tuple,frozen set.
```

```python
# type () :- type function is a pre define function that return the datatype or a
# a class of particular object.
```

```python
# string 
s=""    # empty string
print(s)
print(type(s))
```

```python
s="hello"
print(s)
print(type(s))
print(s[0])    # indexed
s[0]='z'       # immutable
print(s)
```

```python
# list-- collection of elements.
# denoted by[] with csv.
# ordered ,indexed,mutable,heterogenous nature.
l=[]
print(1)
print(type(l))
```

```python
l=[10,20,30,10,"hello",10.5]
print(1)
print(type(l))
print(l[-1])
l[0]=100
print(l)
```

```python
# tuple()---collection of elements.
# denoted by () csv
# ordered,indexed,immutable and heterogenous nature.
t=()
print(t)
print(type(t))
```

```python
# single valued tuple
t=(10,)
print(t)
print(type(t))
```

```python
t=(10,20,30,10,"hello",10.5)
print(t)
print(type(t))
print(t[0])      # indexed
t[0]=100         # immutabe
print(t)
```

```python
# dictionary--   collection of elements
#denoted by {k:v},key must be unique but value can be duplicate.
#ordered,unindexed,mutable
d={}
print(d)
print(type(d))


```

```python
d={"name":"abc","subject":"da","fees":"30000"}
print(d)
print(type(d))
print(d["name"])
d["name"]="pqr"
print(d)
```

```python
d={"name":"abc","subject":"da","fees":30000}
print(d)
print(type(d))
d["Name"]="pqr"
print(d)
```

```python
d={}
print(d)
d["name"]="abc"
print(d)
```

```python
d={"name":["abc","pqr","da"],"subject":["da","da","de"]}
print(d)
print(d["name"],[-1])
d["name"][-1]="xyz"
print(d)
```

```python
# set--collection of element 
# denoted by {} csv.
# stored only unique value.
# unorderde ,unindexed,mutable.
s=set()      #set()-set constructor (empty set)
print(s)
print(type(s))
```

```python
s={10,20,10,20,"hello",10.5,10.5,100}
print(s)
print(type(s))
s.add(2000)    # mutable
print(s)
print(type(s))
```

```python
# frozen set--  collection of unique elements.
# denoted by a constructor.
# unordered,immutable,unindexed and heterogenous
f=frozenset()
print(type(f))
```

```python
f=frozenset({10,20,30,10,20,"hello",10.5})
print(f)
print(type(f))
```
