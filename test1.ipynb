```python
## 1. Find the sum of digits of a number.
num=int(input("enter any number: "))
sum=0
while num!=0:
    dig=num%10
    sum=sum+dig
    num=num//10
print(sum)
```

```python
# Find the  product of digits.
num=int(input("enter any number: "))
pro=1
while num!=0:
    dig=num%10
    pro=pro*dig
    num=num//10
print(pro)
```

```python
# Find the largest digit.
n=int(input("enter any number: "))
max=0
while n!=0:
    dig=n%10
    if dig>max:
        max=dig
    n=n//10
print(max)
```

```python
 # Find the smallest digit.
n=int(input("enter any number: "))
min=9
while n!=0:
    dig=n%10
    if dig<min:
        min=dig
    n=n//10
print(min)
```
